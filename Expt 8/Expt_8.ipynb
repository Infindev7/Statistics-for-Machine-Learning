# Experiment 8: Clustering and Dimensionality Reduction for Pattern Discovery
**Aim:** To apply clustering and dimensionality reduction techniques for unsupervised pattern discovery.

```python
import warnings
warnings.filterwarnings("ignore")
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
plt.rcParams["figure.dpi"] = 100
sns.set_style("whitegrid")
COLS = ["Pregnancies","Glucose","BloodPressure","SkinThickness",
        "Insulin","BMI","DiabetesPedigreeFunction","Age","Outcome"]
df = pd.read_csv("pima-indians-diabetes.csv", names=COLS)
print("Dataset shape:", df.shape)
df.head()
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
```

```python
# Standardize and reduce to 2 principal components
Xs = StandardScaler().fit_transform(df[COLS[:-1]])
pca = PCA(n_components=2, random_state=42)
Z = pca.fit_transform(Xs)
print("Explained variance ratio:", np.round(pca.explained_variance_ratio_, 3))
print("Total explained by 2 PCs:", round(pca.explained_variance_ratio_.sum(), 3))
plt.figure(figsize=(6,5))
plt.scatter(Z[:,0], Z[:,1], c=df["Outcome"], cmap="coolwarm", alpha=0.6)
plt.xlabel("PC1"); plt.ylabel("PC2"); plt.title("PCA Projection colored by Outcome")
plt.tight_layout(); plt.savefig("images/fig1_pca.png", dpi=130, bbox_inches="tight"); plt.show()
```

```python
# Elbow method to choose k
inertias = [KMeans(n_clusters=k, n_init=10, random_state=42).fit(Xs).inertia_ for k in range(1, 7)]
plt.figure(figsize=(6,4)); plt.plot(range(1,7), inertias, marker="o")
plt.xlabel("k"); plt.ylabel("Inertia"); plt.title("Elbow Method")
plt.tight_layout(); plt.savefig("images/fig2_elbow.png", dpi=130, bbox_inches="tight"); plt.show()
```

```python
# KMeans with k=2 on PCA projection; compare with true Outcome
km = KMeans(n_clusters=2, n_init=10, random_state=42).fit(Xs)
plt.figure(figsize=(6,5))
plt.scatter(Z[:,0], Z[:,1], c=km.labels_, cmap="Set2", alpha=0.6)
plt.xlabel("PC1"); plt.ylabel("PC2"); plt.title("KMeans Clusters (k=2) on PCA Projection")
plt.tight_layout(); plt.savefig("images/fig3_kmeans.png", dpi=130, bbox_inches="tight"); plt.show()
print(pd.crosstab(km.labels_, df["Outcome"], rownames=["Cluster"], colnames=["Outcome"]))
```

## Key Observations
- Two principal components capture the dominant variance structure.
- KMeans clusters align meaningfully (though imperfectly) with the true diabetes outcome.
- PCA visualization reveals overlap between groups, explaining classification difficulty.
